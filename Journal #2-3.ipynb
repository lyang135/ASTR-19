```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
def model_exp(x, a, b):
    return a*np.exp(-b*x)

def model_power(x, a, b):
    return a/(x**b)

x_data = np.linspace(1, 10, 20)
y_data = 5*np.exp(-0.5*x_data) + np.random.normal(0, 0.1, len(x_data))

popt_exp, _ = curve_fit(model_exp, x_data, y_data, p0=[1, 1])
popt_power, _ = curve_fit(model_power, x_data, y_data, p0=[1, 1])
print(f"Exponential Model Best Fit: a = {popt_exp[0]:.4f}, b = {popt_exp[1]:.4f}")
print(f"Power Law Model Best Fit: a = {popt_power[0]:.4f}, b = {popt_power[1]:.4f}")

plt.figure(figsize=(10, 6))
plt.scatter(x_data, y_data, label='Original Data', color='black', alpha=0.6)

x_fit = np.linspace(min(x_data), max(x_data), 100)
plt.plot(x_fit, model_exp(x_fit, *popt_exp), label='Exponential Fit', color='blue', linewidth=2)
plt.plot(x_fit, model_power(x_fit, *popt_power), label='Power Law Fit', color='red', linestyle='--', linewidth=2)

plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Curve Fitting Comparison')
plt.legend()
plt.grid(True)

plt.savefig('curve_fit_results.pdf')
plt.show()
```
