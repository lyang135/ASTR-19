```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2*np.pi, 100)
def exponential_decay(x, a, b):
    return a * np.exp(-b * x)

a = 5.0
b = 0.5
```

```python
y = exponential_decay(x, a, b)
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'${a}e^{{-{b}x}}$')

plt.title('Exponential Decay Function')
plt.xlabel('x ($0$ to $2\pi$)')
plt.ylabel('y')
plt.grid(True)
plt.legend()

plt.savefig('exponential_plot.pdf')
plt.show()
```
