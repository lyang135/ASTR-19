```python
import numpy as np

def get_sin(x):
    return np.sin(x)

def get_cos(x):
    return np.cos(x)
```

compute the function sine and cosine of an input x

```python
x_values = np.linspace(0, 2 * np.pi, 1000)

sin_values = get_sin(x_values)
cos_values = get_cos(x_values)
```

```python
print(f"{'x':<10},{'sin(x)':<10},{'cos(x)':<10}")
print("-" * 35)
for i in range(10):
    print(f"{x_values[i]:.4f} | {sin_values[i]:.4f} | {cos_values[i]:.4f}")
```
