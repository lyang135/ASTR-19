```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
x = np.linspace(0, 10, 50)
y = 2*x+5

sigmas = np.linspace(2, 0.2, len(x))
noise = np.random.normal(loc=0, scale=sigmas)
y_noisy = y + noise

plt.figure(figsize=(10, 6))
plt.errorbar(x, y_noisy, yerr=sigmas, fmt='o', color='purple', ecolor='lightgray', elinewidth=1, capsize=2, label='Data with Variable Noise')
plt.plot(x, y, color='black', label='True Relationship', linewidth=2)

plt.xlabel('X (Independent Variable)')
plt.ylabel('Y (Dependent Variable with noise)')
plt.title('Uncertainty Decreases as X Increases')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig('scatter_plot_decreasing_sigma.pdf')
plt.show()
```
