```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.table import Table
from scipy.optimize import curve_fit

data= Table.read("Planet_Lightcurve.fits", format='fits')
print(data)

plt.rcParams['figure.figsize'] = (10,6)
plt.rcParams['font.size'] = 13
```

```python
time = np.array(data['time [days]'], dtype=float)
flux = np.array(data['flux'], dtype=float)

mask = np.isfinite(time) & np.isfinite(flux)

time = time[mask]
flux = flux[mask]

#Experimental uncertainty
sigma_flux = 0.003
```

```python
#Initial Lightcurve Plot
plt.figure()

plt.errorbar(time,
    flux,
    yerr=sigma_flux,
    fmt='.',
    alpha=0.7,
    label='Observed Data')

plt.xlabel("Time [days]")
plt.ylabel("Relative Flux")
plt.title("Raw Exoplanet Lightcurve")
plt.legend()

plt.tight_layout()
plt.savefig("raw_lightcurve.pdf")

plt.show()
```

```python
def artifact_model(t, c0, m, A, f, phi):
    """
    c0   : baseline offset
    m    : linear slope
    A    : oscillation amplitude
    f    : frequency (cycles/day)
    phi  : phase
    """
    return c0 + m*t + A*np.sin(2*np.pi*f*t + phi)
```

```python
#initial parameter guesses
c0_guess = np.mean(flux)
m_guess = 0.0
A_guess = (np.max(flux) - np.min(flux))/4
f_guess = 5.0
phi_guess = 0
p0 = [c0_guess, m_guess, A_guess, f_guess, phi_guess]
```

```python
popt, pcov = curve_fit(
    artifact_model,
    time,
    flux,
    p0=p0,
    sigma=np.full(len(flux), sigma_flux),
    absolute_sigma=True,
    maxfev=100000
)

perr = np.sqrt(np.diag(pcov))
```

```python
names = [
    "Offset c0",
    "Slope m",
    "Amplitude A",
    "Frequency f",
    "Phase phi"
]

print("Best-fit parameters:\n")

for n, val, err in zip(names, popt, perr):
    print(f"{n:15s} = {val:.6f} ± {err:.6f}")

artifact_fit = artifact_model(time, *popt)
```

```python
#plot raw data and model
plt.figure(figsize=(10,6))

plt.errorbar(
    time,
    flux,
    yerr=sigma_flux,
    fmt='.',
    ms=3,
    label='Data'
)

plt.plot(
    time,
    artifact_fit,
    'r',
    lw=2,
    label='Best-fit artifact model'
)

plt.xlabel("Time (days)")
plt.ylabel("Relative Flux")
plt.title("Light Curve and Artifact Model")
plt.legend()

plt.tight_layout()
plt.savefig("artifact_fit.pdf")
plt.show()

#remove artifacts
clean_flux = flux / artifact_fit
clean_flux /= np.median(clean_flux)
```

```python
#plot cleaned light curve
plt.figure(figsize=(10,6))

plt.plot(
    time,
    clean_flux,
    '.',
    ms=3
)

plt.xlabel("Time (days)")
plt.ylabel("Normalized Flux")
plt.title("Artifact-Corrected Light Curve")

plt.tight_layout()
plt.savefig("cleaned_lightcurve.pdf")
plt.show()
```

```python
baseline = np.median(clean_flux)

# transit depth
minimum_flux = np.min(clean_flux)
depth = baseline - minimum_flux

print("Transit depth:")
print(f"D = {depth:.6f}")
```

```python
threshold = baseline - depth/2
in_transit = clean_flux < threshold
indices = np.where(in_transit)[0]

groups = np.split(
    indices,
    np.where(np.diff(indices) > 1)[0] + 1
)

largest_group = max(groups, key=len)

start_index = largest_group[0]
end_index = largest_group[-1]

t_start = time[start_index]
t_end = time[end_index]

duration = t_end - t_start

print(f"Transit duration = {duration:.6f} days")
```

```python
# Planet radius estimate
R_star = 1.0     # solar radii
R_planet = np.sqrt(depth) * R_star

print(f"\nPlanet radius = {R_planet:.6f} R_sun")

# Earth radii conversion
Rsun_to_Rearth = 109.1
R_planet_earth = R_planet * Rsun_to_Rearth

print(f"Planet radius = {R_planet_earth:.2f} R_earth")
```

```python
# Plot transit region
plt.figure(figsize=(10,6))
plt.plot(time, clean_flux, '.', ms=3)
plt.axvline(t_start, color='r', ls='--',
            label='Transit Start')
plt.axvline(t_end, color='g', ls='--',
            label='Transit End')
plt.xlabel("Time (days)")
plt.ylabel("Normalized Flux")
plt.title("Transit Identification")
plt.legend()
plt.tight_layout()
plt.savefig("transit_region.pdf")
plt.show()
```
